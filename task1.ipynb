```python
import pandas as pd
import numpy as np

df = pd.read_excel(
    "British Airways Summer Schedule Dataset - Forage Data Science Task 1.xlsx"
)

print(df.head())
print(df.shape)
print(df.info())
```

```python
df["TOTAL_SEATS"] = (
    df["FIRST_CLASS_SEATS"]
    + df["BUSINESS_CLASS_SEATS"]
    + df["ECONOMY_SEATS"]
)
```

```python
grouped = df.groupby(
    ["TIME_OF_DAY", "HAUL"]
).agg(
    FLIGHT_COUNT=("FLIGHT_NO", "count"),
    TOTAL_SEATS=("TOTAL_SEATS", "sum"),
    TIER1=("TIER1_ELIGIBLE_PAX", "sum"),
    TIER2=("TIER2_ELIGIBLE_PAX", "sum"),
    TIER3=("TIER3_ELIGIBLE_PAX", "sum")
).reset_index()
```

```python
grouped["TIER1_PERCENT"] = (
    grouped["TIER1"] / grouped["TOTAL_SEATS"] * 100
)

grouped["TIER2_PERCENT"] = (
    grouped["TIER2"] / grouped["TOTAL_SEATS"] * 100
)

grouped["TIER3_PERCENT"] = (
    grouped["TIER3"] / grouped["TOTAL_SEATS"] * 100
)

print(grouped.round(2))
```

```python
grouped.to_excel(
    "BA_Lounge_Eligibility_Lookup.xlsx",
    index=False
)
```
